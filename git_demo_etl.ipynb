```python
# Git Demo ETL Notebook
# 本 Notebook 演示从 Git Folder 运行 ETL 任务

from pyspark.sql import functions as F
from datetime import datetime

print(f"ETL Job 开始执行: {datetime.now()}")
print(f"运行环境: Databricks Git Folder Demo")
```

```python
# Step 1: 创建示例销售数据
data = [
    ("2026-10-01", "产品A", 100, 29.99),
    ("2026-10-01", "产品B", 50, 49.99),
    ("2026-10-02", "产品A", 120, 29.99),
    ("2026-10-02", "产品C", 80, 19.99),
    ("2026-10-03", "产品B", 60, 49.99),
    ("2026-10-03", "产品C", 90, 19.99),
]

df = spark.createDataFrame(data, ["sale_date", "product", "quantity", "unit_price"])
df.show()
```

```python
# Step 2: 数据转换 - 按产品汇总销售额
result = (
    df.withColumn("total_amount", F.col("quantity") * F.col("unit_price"))
    .groupBy("product")
    .agg(
        F.sum("quantity").alias("total_quantity"),
        F.round(F.sum("total_amount"), 2).alias("total_revenue")
    )
    .orderBy(F.desc("total_revenue"))
)

print("=== 产品销售汇总 ===")
result.show()
```

```python
# Step 3: 输出完成日志
print(f"ETL Job 执行完成: {datetime.now()}")
print(f"处理记录数: {df.count()}")
print(f"输出记录数: {result.count()}")
```
